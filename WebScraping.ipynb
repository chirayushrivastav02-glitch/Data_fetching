```python
import requests

url = "https://yahoo-finance15.p.rapidapi.com/api/v2/markets/stock/history"

querystring = {"symbol":"AAPL","interval":"1m","limit":"640"}

headers = {
	"x-rapidapi-key": "5c552018f9mshbaa4673dceeedd5p15317bjsn40c4553477e2",
	"x-rapidapi-host": "yahoo-finance15.p.rapidapi.com",
	"Content-Type": "application/json"
}

response = requests.get(url, headers=headers, params=querystring)

data = response.json()

df = pd.DataFrame(data['body'])
df.head()
```

```python
import matplotlib.dates as mdates

df["timestamp"] = pd.to_datetime(df["timestamp"])

ax = df.plot(
    x="timestamp",
    y=["open",
       "moving_average for 5 minutes",
       "moving_average for 20 minutes",
       "moving_average for 50 minutes"],
    figsize=(15, 6)
)

ax.xaxis.set_major_locator(mdates.HourLocator(interval=6))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%d-%m %H:%M"))

plt.grid()
plt.show()
```
